```python
from pathlib import Path
    import pandas as pd
    DATA = Path("../04_Data_Cleaning/cleaned_data")
    emp = pd.read_csv(DATA/"employees.csv")
    rec = pd.read_csv(DATA/"recruitment_pipeline.csv")
    train = pd.read_csv(DATA/"training_records.csv")
```

```python
dept = emp.groupby("department").agg(
        headcount=("employee_id","count"), exits=("attrition_flag","sum"),
        attrition_rate=("attrition_flag","mean"),
        satisfaction=("satisfaction_score","mean"),
        engagement=("engagement_score","mean")).reset_index()
    dept["attrition_rate_pct"] = (dept.attrition_rate*100).round(2)
    display(dept.sort_values("attrition_rate",ascending=False))
```

```python
source = rec.groupby("candidate_source").agg(
        applications=("application_id","count"),
        hires=("current_stage",lambda x:(x=="Hired").sum())).reset_index()
    source["conversion_pct"]=(source.hires/source.applications*100).round(2)
    display(source.sort_values("conversion_pct",ascending=False))
```

```python
learning = train.groupby("course_category").agg(
        enrollments=("training_id","count"),
        completion_pct=("completion_status",lambda x:(x=="Completed").mean()*100),
        mean_score=("assessment_score","mean")).reset_index()
    display(learning.sort_values("completion_pct",ascending=False))
```

## Executive narrative template
    **Finding:** State the measured pattern, segment and period.

    **Evidence:** Include numerator, denominator, rate and comparison baseline.

    **Business implication:** Explain the potential workforce or operating impact without claiming causation.

    **Recommendation:** Propose a measurable pilot, owner and review date.

    **Success metric:** Define how improvement will be evaluated (e.g. regretted attrition, 90-day retention, training completion or time-to-hire).
