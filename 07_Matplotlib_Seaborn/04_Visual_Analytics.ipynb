# 04 — Matplotlib & Seaborn Visual Analytics
    Create reusable, presentation-ready charts from the cleaned data.

```python
from pathlib import Path
    import pandas as pd
    import matplotlib.pyplot as plt
    import seaborn as sns
    sns.set_theme(style="whitegrid", context="talk")
    DATA = Path("../04_Data_Cleaning/cleaned_data")
    OUT = Path("../07_Matplotlib_Seaborn/charts")
    OUT.mkdir(exist_ok=True)
    emp = pd.read_csv(DATA/"employees.csv")
    train = pd.read_csv(DATA/"training_records.csv")
    perf = pd.read_csv(DATA/"performance_reviews.csv")
```

```python
# Department attrition rate
    summary = emp.groupby("department").attrition_flag.mean().mul(100).sort_values()
    fig, ax = plt.subplots(figsize=(12,6))
    sns.barplot(x=summary.values, y=summary.index, ax=ax)
    ax.set(title="Attrition Rate by Department", xlabel="Attrition rate (%)", ylabel="")
    fig.tight_layout()
    fig.savefig(OUT/"attrition_by_department.png", dpi=180)
    plt.show()
```

```python
# Salary distribution by job level
    fig, ax = plt.subplots(figsize=(12,6))
    sns.boxplot(data=emp, x="job_level", y="annual_salary_inr", ax=ax)
    ax.set(title="Salary Distribution by Job Level", xlabel="Job level", ylabel="Annual salary (INR)")
    fig.tight_layout()
    fig.savefig(OUT/"salary_by_level.png", dpi=180)
    plt.show()
```

```python
# Correlation heatmap for numeric workforce measures
    numeric = emp[["age","tenure_years","annual_salary_inr","satisfaction_score",
                   "engagement_score","attrition_flag"]].corr()
    fig, ax = plt.subplots(figsize=(10,7))
    sns.heatmap(numeric, annot=True, cmap="vlag", center=0, fmt=".2f", ax=ax)
    ax.set_title("Workforce Variable Correlations")
    fig.tight_layout()
    fig.savefig(OUT/"workforce_correlation_heatmap.png", dpi=180)
    plt.show()
```

```python
# Performance rating distribution
    fig, ax = plt.subplots(figsize=(10,5))
    sns.histplot(data=perf, x="performance_rating", bins=20, kde=True, ax=ax)
    ax.set(title="Performance Rating Distribution", xlabel="Performance rating")
    fig.tight_layout()
    fig.savefig(OUT/"performance_rating_distribution.png", dpi=180)
    plt.show()
```
