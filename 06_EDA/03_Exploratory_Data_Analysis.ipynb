# 03 — Exploratory Data Analysis (EDA)
    Analyze workforce size, attrition, compensation, attendance, learning and recruitment.

```python
from pathlib import Path
    import pandas as pd
    import numpy as np
    import matplotlib.pyplot as plt
    import seaborn as sns
    DATA = Path("../04_Data_Cleaning/cleaned_data")
    emp = pd.read_csv(DATA/"employees.csv")
    perf = pd.read_csv(DATA/"performance_reviews.csv")
    train = pd.read_csv(DATA/"training_records.csv")
    att = pd.read_csv(DATA/"monthly_attendance.csv")
    rec = pd.read_csv(DATA/"recruitment_pipeline.csv")
    print("Employees:", emp.shape, "Performance:", perf.shape, "Training:", train.shape)
```

```python
kpis = {
        "headcount": len(emp),
        "active_employees": int((emp.employment_status=="Active").sum()),
        "exited_employees": int((emp.employment_status=="Exited").sum()),
        "attrition_rate_pct": round(emp.attrition_flag.mean()*100,2),
        "average_tenure_years": round(emp.tenure_years.mean(),2),
        "average_salary_inr": round(emp.annual_salary_inr.mean(),0)
    }
    kpis
```

```python
dept_attrition = emp.groupby("department").agg(
        employees=("employee_id","count"), exits=("attrition_flag","sum"),
        attrition_rate=("attrition_flag","mean")).sort_values("attrition_rate",ascending=False)
    dept_attrition["attrition_rate_pct"] = dept_attrition["attrition_rate"]*100
    display(dept_attrition)
```

```python
salary_summary = emp.groupby(["department","job_level"]).annual_salary_inr.agg(
        ["count","mean","median","min","max"]).round(0)
    display(salary_summary)
```

```python
monthly_absence = att.groupby("month",as_index=False).agg(
        avg_absent_days=("days_absent","mean"), avg_late_arrivals=("late_arrivals","mean"))
    display(monthly_absence.head())
```
