# 01 — HR Data Import & Profiling
    Load the source CSV files, inspect schemas, profile row counts and establish keys.

```python
from pathlib import Path
    import pandas as pd
    DATA = Path("../01_Raw_Data")
    employees = pd.read_csv(DATA/"employees.csv", parse_dates=["hire_date"])
    training = pd.read_csv(DATA/"training_records.csv", parse_dates=["enrollment_date"])
    performance = pd.read_csv(DATA/"performance_reviews.csv")
    attendance = pd.read_csv(DATA/"monthly_attendance.csv")
    recruitment = pd.read_csv(DATA/"recruitment_pipeline.csv", parse_dates=["application_date"])
    datasets = {"employees":employees, "training":training, "performance":performance,
                "attendance":attendance, "recruitment":recruitment}
    {name: df.shape for name, df in datasets.items()}
```

```python
for name, df in datasets.items():
        print(f"\\n{name.upper()}")
        display(df.head())
        display(df.info())
```

## Initial data-quality scan
    Review null counts, duplicates and unique key counts before cleaning.

```python
profile = pd.DataFrame({
        "rows": {k: len(v) for k,v in datasets.items()},
        "columns": {k: v.shape[1] for k,v in datasets.items()},
        "duplicate_rows": {k: int(v.duplicated().sum()) for k,v in datasets.items()},
        "missing_cells": {k: int(v.isna().sum().sum()) for k,v in datasets.items()}
    })
    display(profile)
```
